```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score,f1_score,classification_report,precision_score,recall_score
```

```python
feature_cols = [
    'eda_mean', 'eda_std', 'eda_min', 'eda_max', 'eda_range', 'eda_slope',
    'temp_mean', 'temp_std', 'temp_min', 'temp_max', 'temp_range', 'temp_slope',
    'hr_mean', 'hr_std', 'hr_min', 'hr_max',
    'ibi_mean', 'ibi_sdnn', 'ibi_rmssd', 'ibi_pnn50'
]
X = df[feature_cols]
y = df['binary_label']
```

```python
logo = LeaveOneGroupOut()
```

# Logistic Regression

```python
accuracies = []
precisions = []
recalls = []
f1s = []
all_y_true = []
all_y_pred = []

for train_idx, test_idx in logo.split(X, y, groups=df["subject_id"]):

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    lr_model = LogisticRegression(max_iter=5000, random_state=42)

    lr_model.fit(X_train, y_train)

    y_pred = lr_model.predict(X_test)

    accuracies.append(accuracy_score(y_test, y_pred))
    precisions.append(precision_score(y_test, y_pred, zero_division=0))
    recalls.append(recall_score(y_test, y_pred, zero_division=0))
    f1s.append(f1_score(y_test, y_pred, zero_division=0))
    
    all_y_true.extend(y_test)
    all_y_pred.extend(y_pred)

print("Average LOSO Accuracy:", np.mean(accuracies))
print("Average LOSO Precision:", np.mean(precisions))
print("Average LOSO Recall:", np.mean(recalls))
print("Average LOSO F1-Score:", np.mean(f1s))
print("\nClassification Report (Aggregated):\n", classification_report(all_y_true, all_y_pred))
```

# Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

logo = LeaveOneGroupOut()

accuracies = []
precisions = []
recalls = []
f1s = []
all_y_true = []
all_y_pred = []

for train_idx, test_idx in logo.split(X, y, groups=df["subject_id"]):

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    clf_model = DecisionTreeClassifier(
    max_depth=6,
    random_state=42,
    )

    clf_model.fit(X_train, y_train)

    y_pred = clf_model.predict(X_test)

    accuracies.append(accuracy_score(y_test, y_pred))
    precisions.append(precision_score(y_test, y_pred, zero_division=0))
    recalls.append(recall_score(y_test, y_pred, zero_division=0))
    f1s.append(f1_score(y_test, y_pred, zero_division=0))
    
    all_y_true.extend(y_test)
    all_y_pred.extend(y_pred)

print("Average LOSO Accuracy:", np.mean(accuracies))
print("Average LOSO Precision:", np.mean(precisions))
print("Average LOSO Recall:", np.mean(recalls))
print("Average LOSO F1-Score:", np.mean(f1s))
print("\nClassification Report (Aggregated):\n", classification_report(all_y_true, all_y_pred))
```

```python
from xgboost import XGBClassifier
logo = LeaveOneGroupOut()

accuracies = []
precisions = []
recalls = []
f1s = []
all_y_true = []
all_y_pred = []

for train_idx, test_idx in logo.split(X, y, groups=df["subject_id"]):

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    model = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=35
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracies.append(accuracy_score(y_test, y_pred))
    precisions.append(precision_score(y_test, y_pred, zero_division=0))
    recalls.append(recall_score(y_test, y_pred, zero_division=0))
    f1s.append(f1_score(y_test, y_pred, zero_division=0))
    
    all_y_true.extend(y_test)
    all_y_pred.extend(y_pred)

print("Average LOSO Accuracy:", np.mean(accuracies))
print("Average LOSO Precision:", np.mean(precisions))
print("Average LOSO Recall:", np.mean(recalls))
print("Average LOSO F1-Score:", np.mean(f1s))
print("\nClassification Report (Aggregated):\n", classification_report(all_y_true, all_y_pred))
```
