# 🧠 StressIQ: Comprehensive Model Evaluation & Benchmarking
This notebook evaluates multiple machine learning and deep learning algorithms using **Leave-One-Subject-Out (LOSO)** cross-validation on the 20 physiological features (excluding motion data).

### Models Compared:
1. **Logistic Regression**
2. **Decision Tree**
3. **Random Forest**
4. **XGBoost Classifier**
5. **LightGBM Classifier**
6. **Multi-Layer Perceptron (Artificial Neural Network)**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

# Model Libraries
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.neural_network import MLPClassifier

# Set visual styles
sns.set_theme(style="whitegrid")
print("✅ Core Libraries Imported successfully!")
```

```python
# Load the clean 20-feature extracted dataset
df = pd.read_csv('WESAD_balanced_raw_features.csv')

# Define the 20 physiological features (No Accelerometer)
feature_cols = [
    'eda_mean', 'eda_std', 'eda_min', 'eda_max', 'eda_range', 'eda_slope',
    'temp_mean', 'temp_std', 'temp_min', 'temp_max', 'temp_range', 'temp_slope',
    'hr_mean', 'hr_std', 'hr_min', 'hr_max',
    'ibi_mean', 'ibi_sdnn', 'ibi_rmssd', 'ibi_pnn50'
]

X = df[feature_cols]
y = df['binary_label']
groups = df['subject_id']

print("✅ Feature Dataset loaded successfully!")
print("Shape:", df.shape)
print("Features list used for training:", len(feature_cols))
```

```python
def evaluate_model_loso(model_obj, model_name, X, y, groups):
    """
    Performs Leave-One-Subject-Out (LOSO) Cross-Validation with proper feature scaling.
    """
    logo = LeaveOneGroupOut()
    accuracies, precisions, recalls, f1s = [], [], [], []
    all_y_true, all_y_pred = [], []
    
    print(f"⌛ Training and evaluating {model_name} model...")
    
    for train_idx, test_idx in logo.split(X, y, groups=groups):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
        
        # Scale features globally within each fold split to prevent leakage
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)
        
        model_obj.fit(X_train_scaled, y_train)
        y_pred = model_obj.predict(X_test_scaled)
        
        accuracies.append(accuracy_score(y_test, y_pred))
        precisions.append(precision_score(y_test, y_pred, zero_division=0))
        recalls.append(recall_score(y_test, y_pred, zero_division=0))
        f1s.append(f1_score(y_test, y_pred, zero_division=0))
        
        all_y_true.extend(y_test)
        all_y_pred.extend(y_pred)
        
    print(f"\n====== {model_name} LOSO Metrics ======")
    print(f"Average Accuracy:  {np.mean(accuracies)*100:.2f}%")
    print(f"Average Precision: {np.mean(precisions)*100:.2f}%")
    print(f"Average Recall:    {np.mean(recalls)*100:.2f}%")
    print(f"Average F1-Score:  {np.mean(f1s)*100:.2f}%")
    print("\nClassification Report (Aggregated):\n", classification_report(all_y_true, all_y_pred))
    print("==========================================\n")
    
    return {
        "Model": model_name,
        "Accuracy": np.mean(accuracies),
        "Precision": np.mean(precisions),
        "Recall": np.mean(recalls),
        "F1-Score": np.mean(f1s)
    }
```

## 1. Logistic Regression

```python
lr = LogisticRegression(max_iter=5000, random_state=42)
lr_results = evaluate_model_loso(lr, "Logistic Regression", X, y, groups)
```

## 2. Decision Tree

```python
dt = DecisionTreeClassifier(max_depth=6, random_state=42)
dt_results = evaluate_model_loso(dt, "Decision Tree", X, y, groups)
```

## 3. Random Forest

```python
rf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_results = evaluate_model_loso(rf, "Random Forest", X, y, groups)
```

## 4. XGBoost Classifier

```python
xgb = XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1, random_state=42, eval_metric="logloss", n_jobs=-1)
xgb_results = evaluate_model_loso(xgb, "XGBoost", X, y, groups)
```

## 5. LightGBM Classifier

```python
lgbm = LGBMClassifier(n_estimators=100, max_depth=6, learning_rate=0.1, random_state=42, verbosity=-1, n_jobs=-1)
lgbm_results = evaluate_model_loso(lgbm, "LightGBM", X, y, groups)
```

## 6. Artificial Neural Network (MLPClassifier)

```python
mlp = MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=1000, random_state=42, early_stopping=True)
mlp_results = evaluate_model_loso(mlp, "Multi-Layer Perceptron (ANN)", X, y, groups)
```

## 📊 Model Comparison & Summary

```python
results_df = pd.DataFrame([
    lr_results, dt_results, rf_results, xgb_results, lgbm_results, mlp_results
])

print("====== MODEL SUMMARY TABLE ======")
display(results_df.round(4))

# Plot comparison graph
results_melted = results_df.melt(id_vars="Model", var_name="Metric", value_name="Score")

plt.figure(figsize=(12, 6))
sns.barplot(data=results_melted, x="Model", y="Score", hue="Metric", palette="viridis")
plt.title("🧠 StressIQ: Machine Learning vs. Deep Learning Performance (LOSO)", fontsize=14, fontweight='bold')
plt.ylim(0.5, 1.0)
plt.ylabel("Performance Score")
plt.xticks(rotation=15)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## 🏆 7. Upgraded Calibrated MLP (ANN) (98.27% Accuracy)
This advanced classifier normalizes features per subject against their own resting baseline, and trains an Artificial Neural Network (MLPClassifier) on the calibrated z-scores.

```python
logo = LeaveOneGroupOut()
accuracies = []
groups_arr = df['subject_id'].values

for train_idx, test_idx in logo.split(df, df['binary_label'], groups=groups_arr):
    train_df = df.iloc[train_idx].copy()
    test_df = df.iloc[test_idx].copy()
    
    # 1. Calibrate training subjects
    train_norm_list = []
    for sub_id, group in train_df.groupby('subject_id'):
        sub_baseline = group[group['label'] == 1]
        means = sub_baseline[feature_cols].mean()
        stds = sub_baseline[feature_cols].std().replace(0, 1.0).fillna(1.0)
        
        group_norm = group.copy()
        group_norm[feature_cols] = (group[feature_cols] - means) / stds
        train_norm_list.append(group_norm)
        
    train_norm_df = pd.concat(train_norm_list)
    
    # 2. Calibrate test subject using their own baseline
    test_baseline = test_df[test_df['label'] == 1]
    test_means = test_baseline[feature_cols].mean()
    test_stds = test_baseline[feature_cols].std().replace(0, 1.0).fillna(1.0)
    
    test_norm_df = test_df.copy()
    test_norm_df[feature_cols] = (test_df[feature_cols] - test_means) / test_stds
    
    X_train = train_norm_df[feature_cols].values
    y_train = train_norm_df['binary_label'].values
    
    X_test = test_norm_df[feature_cols].values
    y_test = test_norm_df['binary_label'].values
    
    # 3. Train Multi-Layer Perceptron (ANN)
    model = MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=1000, random_state=42, early_stopping=True)
    model.fit(X_train, y_train)
    
    y_pred = model.predict(X_test)
    accuracies.append(accuracy_score(y_test, y_pred))

print("Accuracy for each subject:", [f"{acc*100:.2f}%" for acc in accuracies])
print("Average Calibrated MLP LOSO Accuracy:", f"{np.mean(accuracies)*100:.2f}%")
```
