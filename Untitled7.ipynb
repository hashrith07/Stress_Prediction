```python
feature_cols = [
    'eda_mean', 'eda_std', 'eda_min', 'eda_max', 'eda_range', 'eda_slope',
    'temp_mean', 'temp_std', 'temp_min', 'temp_max', 'temp_range', 'temp_slope',
    'hr_mean', 'hr_std', 'hr_min', 'hr_max',
    'ibi_mean', 'ibi_sdnn', 'ibi_rmssd', 'ibi_pnn50'
]
```

```python
X=df[feature_cols]
y=df['binary_label']
```

```python
groups=df['subject_id']
```

```python
logo = LeaveOneGroupOut()
```

```python
def evaluate_model_loso(model_obj, model_name, X, y, groups):
    logo = LeaveOneGroupOut()
    accuracies, precisions, recalls, f1s = [], [], [], []
    all_y_true, all_y_pred = [], []
    
    print(f"⌛ Training and evaluating {model_name}...")
    
    for train_idx, test_idx in logo.split(X, y, groups=groups):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
        
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)
        
        model_obj.fit(X_train_scaled, y_train)
        y_pred = model_obj.predict(X_test_scaled)
        
        accuracies.append(accuracy_score(y_test, y_pred))
        precisions.append(precision_score(y_test, y_pred, zero_division=0))
        recalls.append(recall_score(y_test, y_pred, zero_division=0))
        f1s.append(f1_score(y_test, y_pred, zero_division=0))
        
        all_y_true.extend(y_test)
        all_y_pred.extend(y_pred)
        
    print(f"\n====== {model_name} LOSO Metrics ======")
    print(f"Average Accuracy:  {np.mean(accuracies)*100:.2f}%")
    print(f"Average Precision: {np.mean(precisions)*100:.2f}%")
    print(f"Average Recall:    {np.mean(recalls)*100:.2f}%")
    print(f"Average F1-Score:  {np.mean(f1s)*100:.2f}%")
    print("\nClassification Report (Aggregated):\n", classification_report(all_y_true, all_y_pred))
    print("==========================================\n")
    
    return {
        "Model": model_name,
        "Accuracy": np.mean(accuracies),
        "Precision": np.mean(precisions),
        "Recall": np.mean(recalls),
        "F1-Score": np.mean(f1s)
    }
```

# Logistic Regression

```python
lr = LogisticRegression(max_iter=5000, random_state=42)
lr_results = evaluate_model_loso(lr, "Logistic Regression", X, y, groups)
```

# Decision Tree

```python
dt = DecisionTreeClassifier(max_depth=6, random_state=42)
dt_results = evaluate_model_loso(dt, "Decision Tree", X, y, groups)
```

# Random Forest

```python
rf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_results = evaluate_model_loso(rf, "Random Forest", X, y, groups)
```

# XGBoost Classifier

```python
xgb = XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1, random_state=42, eval_metric="logloss", n_jobs=-1)
xgb_results = evaluate_model_loso(xgb, "XGBoost", X, y, groups)
```

# LightGBM Classifier

```python
lgbm = LGBMClassifier(n_estimators=100, max_depth=6, learning_rate=0.1, random_state=42, verbosity=-1, n_jobs=-1)
lgbm_results = evaluate_model_loso(lgbm, "LightGBM", X, y, groups)
```

# Artificial Neural Network (MLP Classifier)

```python
mlp = MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=1000, random_state=42, early_stopping=True)
mlp_results = evaluate_model_loso(mlp, "Multi-Layer Perceptron (ANN)", X, y, groups)
```

# Results

```python
results_df = pd.DataFrame([
    lr_results, dt_results, rf_results, xgb_results, lgbm_results, mlp_results
])

print("====== MODEL SUMMARY TABLE ======")
display(results_df.round(4))

results_melted = results_df.melt(id_vars="Model", var_name="Metric", value_name="Score")

plt.figure(figsize=(12, 6))
sns.barplot(data=results_melted, x="Model", y="Score", hue="Metric", palette="viridis")
plt.title("🧠 StressIQ: Model Comparison Benchmarks (LOSO)", fontsize=14, fontweight='bold')
plt.ylim(0.5, 1.0)
plt.ylabel("Performance Score")
plt.xticks(rotation=15)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
